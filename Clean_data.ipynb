```python
def remove_emojis(text):
    """
    Удаляет ВСЕ смайлики и эмодзи из текста
    """
    if pd.isna(text):
        return text
    
    # Паттерн для удаления всех смайликов
    emoji_pattern = re.compile("["
                           u"\U0001F600-\U0001F64F"  # эмотиконы
                           u"\U0001F300-\U0001F5FF"  # символы и пиктограммы
                           u"\U0001F680-\U0001F6FF"  # транспорт и карты
                           u"\U0001F1E0-\U0001F1FF"  # флаги (iOS)
                           u"\U00002702-\U000027B0"
                           u"\U000024C2-\U0001F251"
                           "]+", flags=re.UNICODE)
    
    # Удаляем смайлики и чистим пробелы
    text = emoji_pattern.sub(r'', str(text))
    text = re.sub(r'\s+', ' ', text).strip()
    
    return text

def clean_dataframe_emojis(df, text_column='text'):
    """
    Очищает указанный столбец датафрейма от всех смайликов
    """
    df_clean = df.copy()
    df_clean[text_column] = df_clean[text_column].apply(remove_emojis)
    return df_clean
```

```python
df_clean = clean_dataframe_emojis(df)
```

```python
df_clean.to_csv('./data/websites_updated_cleaned.csv',index=False)
```

## Создаем словарь для Chroma

```python
def dataframe_to_websites_dict(df):
    """
    Преобразует датафрейм в словарь формата WEBSITES_DATA
    
    Parameters:
    df (pd.DataFrame): исходный датафрейм со столбцами web_id, url, kind, title, text
    
    Returns:
    list: список словарей в требуемом формате
    """
    WEBSITES_DATA = []
    
    for _, row in df.iterrows():
        # Создаем элемент в нужном формате
        item = {
            "text": str(row['text']) if pd.notna(row['text']) else "",
            "metadata": {
                "web_id": str(row['web_id']) if pd.notna(row['web_id']) else "",
                "url": str(row['url']) if pd.notna(row['url']) else "",
                "kind": str(row['kind']) if pd.notna(row['kind']) else "",
                "title": str(row['title']) if pd.notna(row['title']) else "",
            }
        }
        WEBSITES_DATA.append(item)
    
    return WEBSITES_DATA
```

```python
websites_data = dataframe_to_websites_dict(df_clean)
```

```python
with open('./data/websites_dict.json', 'w', encoding='utf-8') as f:
    json.dump(websites_data, f, ensure_ascii=False, indent=2)
```
